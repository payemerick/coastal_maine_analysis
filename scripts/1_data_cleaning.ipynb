```r
#set working directory
setwd("/content/drive/MyDrive/coastal_maine_analysis")
getwd()

# load libraries
library(readr)
library(dplyr)
library(tidyr)
```

```r
# read in raw data
tow_data_raw <- read.csv("raw_data/MaineDMR_Trawl_Survey_Tow_Data_2026-11-18.csv")
catch_at_length_raw <- read.csv("raw_data/MaineDMR_Trawl_Survey_Catch_at_Length_Data_2026-11-18.csv")
```

```r
glimpse(tow_data_raw)
```

```r
glimpse(catch_at_length_raw)
```

```r
# filter catch_at_length table to just be fish of interest

target_fish <- c("Herring Atlantic", "Sand Lance American", "Hake White")

catch_length_filtered <- catch_at_length_raw %>%
  rename_with(tolower) %>% # make all columns lowercase
  filter(common_name %in% target_fish)

glimpse(catch_length_filtered)
```

```r
# Summarize filtered catch at length by average length, frequency, and total frequency for each fish species
catch_length_summary <- catch_length_filtered %>%
    group_by(common_name) %>%
    summarise(
        mean_length = mean(length, na.rm = TRUE), # remove NAs
        mean_frequency = mean(frequency, na.rm = TRUE),
        total_frequency = sum(frequency, na.rm = TRUE)
    )

catch_length_summary
```

###Save cleaned data

```r
write_csv(catch_length_filtered, "clean_data/catch_length_filtered.csv")
write_csv(catch_length_summary, "clean_data/catch_length_summary.csv")
list.files("clean_data") # check that it's there
```
